```python
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error
import matplotlib.pyplot as plt
```

```python
xgb_df = state_df.copy()
```

```python
xgb_df['lag_1'] = xgb_df['Total'].shift(1)

xgb_df['lag_7'] = xgb_df['Total'].shift(7)

xgb_df['lag_30'] = xgb_df['Total'].shift(30)
```

```python
xgb_df['rolling_mean'] = xgb_df['Total'].rolling(7).mean()

xgb_df['rolling_std'] = xgb_df['Total'].rolling(7).std()
```

```python
xgb_df['day_of_week'] = xgb_df['Date'].dt.dayofweek

xgb_df['month'] = xgb_df['Date'].dt.month
```

```python
xgb_df = xgb_df.dropna()
```

```python
print(xgb_df.head())
```

```python
features = [
    'lag_1',
    'lag_7',
    'lag_30',
    'rolling_mean',
    'rolling_std',
    'day_of_week',
    'month'
]
```

```python
X = xgb_df[features]

y = xgb_df['Total']
```

```python
split_index = int(len(xgb_df) * 0.8)

X_train = X[:split_index]
X_test = X[split_index:]

y_train = y[:split_index]
y_test = y[split_index:]
```

```python
model = XGBRegressor(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=5,
    random_state=42
)

model.fit(X_train, y_train)
```

```python
predictions = model.predict(X_test)
```

```python
mae = mean_absolute_error(y_test, predictions)

print("MAE:", mae)
```

```python
plt.figure(figsize=(12,6))

plt.plot(y_test.values, label='Actual')

plt.plot(predictions, label='Predicted')

plt.legend()

plt.title("XGBoost Forecast")

plt.show()
```

```python
from statsmodels.tsa.arima.model import ARIMA
```

```python
arima_data = state_df['Total']
```

```python
split_index = int(len(arima_data) * 0.8)

train = arima_data[:split_index]

test = arima_data[split_index:]
```

```python
model = ARIMA(train, order=(5,1,0))

model_fit = model.fit()
```

```python
forecast = model_fit.forecast(steps=len(test))
```

```python
from sklearn.metrics import mean_absolute_error

arima_mae = mean_absolute_error(test, forecast)

print("ARIMA MAE:", arima_mae)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

plt.plot(test.values, label='Actual')

plt.plot(forecast.values, label='ARIMA Forecast')

plt.legend()

plt.title("ARIMA Forecast")

plt.show()
```
